**1. Greedy Best-First Search**

```python
import heapq

def h(state, goal):
    return sum(state[i] != goal[i] and state[i] != 0
               for i in range(9))

def neighbors(state):
    result = []
    zero = state.index(0)
    row, col = divmod(zero, 3)

    for dr, dc in [(-1,0), (1,0), (0,-1), (0,1)]:
        r, c = row + dr, col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            new = list(state)
            pos = r * 3 + c
            new[zero], new[pos] = new[pos], new[zero]
            result.append(tuple(new))

    return result

def greedy(start, goal):
    pq = [(h(start, goal), start)]
    visited = set()

    while pq:
        _, state = heapq.heappop(pq)

        if state in visited:
            continue

        visited.add(state)

        print(state[:3])
        print(state[3:6])
        print(state[6:])
        print()

        if state == goal:
            print("Goal Reached!")
            return

        for next_state in neighbors(state):
            heapq.heappush(pq, (h(next_state, goal), next_state))


start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

greedy(start, goal)
```

**2. A* Search**

```python
import heapq

def h(state, goal):
    return sum(state[i] != goal[i] and state[i] != 0
               for i in range(9))

def neighbors(state):
    result = []
    zero = state.index(0)
    row, col = divmod(zero, 3)

    for dr, dc in [(-1,0), (1,0), (0,-1), (0,1)]:
        r, c = row + dr, col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            new = list(state)
            pos = r * 3 + c
            new[zero], new[pos] = new[pos], new[zero]
            result.append(tuple(new))

    return result

def astar(start, goal):
    pq = [(h(start, goal), 0, start)]
    visited = set()

    while pq:
        f, g, state = heapq.heappop(pq)

        if state in visited:
            continue

        visited.add(state)

        print(state[:3])
        print(state[3:6])
        print(state[6:])
        print()

        if state == goal:
            print("Goal Reached!")
            return

        for next_state in neighbors(state):
            new_g = g + 1
            new_f = new_g + h(next_state, goal)

            heapq.heappush(pq, (new_f, new_g, next_state))


start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

astar(start, goal)
```

**3. AO* Search**

```python
graph = {
    'A': [('B', 'C'), ('D',)],
    'B': [('E',)],
    'C': [('F',)],
    'D': [('G',)],
    'E': [],
    'F': [],
    'G': []
}

cost = {
    'A': 4,
    'B': 2,
    'C': 2,
    'D': 5,
    'E': 0,
    'F': 0,
    'G': 0
}

def ao_star(node):

    if not graph[node]:
        return cost[node]

    best = float('inf')

    for option in graph[node]:

        total = 0

        for child in option:
            total += cost[child]

        if total < best:
            best = total

    cost[node] = best
    return best


print("AO* Search")

answer = ao_star('A')

print("Minimum Cost:", answer)
print("Updated Costs:", cost)
```

**4. Memory-Bounded Heuristic Search**

```python
def h(state, goal):
    return sum(state[i] != goal[i] and state[i] != 0
               for i in range(9))


def neighbors(state):
    result = []
    zero = state.index(0)
    row, col = divmod(zero, 3)

    for dr, dc in [(-1,0), (1,0), (0,-1), (0,1)]:
        r, c = row + dr, col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            new = list(state)
            pos = r * 3 + c
            new[zero], new[pos] = new[pos], new[zero]
            result.append(tuple(new))

    return result


def search(path, g, limit, goal):

    state = path[-1]

    f = g + h(state, goal)

    if f > limit:
        return f

    if state == goal:
        return True

    minimum = float('inf')

    for next_state in neighbors(state):

        if next_state not in path:

            path.append(next_state)

            result = search(
                path, g + 1, limit, goal
            )

            if result is True:
                return True

            minimum = min(minimum, result)

            path.pop()

    return minimum


def ida_star(start, goal):

    limit = h(start, goal)
    path = [start]

    while True:

        result = search(path, 0, limit, goal)

        if result is True:
            return path

        limit = result


start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

solution = ida_star(start, goal)

print("IDA* Solution:")

for state in solution:
    print(state[:3])
    print(state[3:6])
    print(state[6:])
    print()
```
