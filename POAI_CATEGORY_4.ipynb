**1. Breadth First Search — BFS**

```python
graph = {
    'A': ['B', 'C'],
    'B': ['D'],
    'C': ['E'],
    'D': [],
    'E': []
}

queue = ['A']
visited = []

while queue:

    node = queue.pop(0)

    if node not in visited:
        print(node, end=" ")
        visited.append(node)

        for x in graph[node]:
            queue.append(x)
```

**2. Depth First Search — DFS**

```python
graph = {
    'A': ['B', 'C'],
    'B': ['D'],
    'C': ['E'],
    'D': [],
    'E': []
}

stack = ['A']

while stack:

    node = stack.pop()
    print(node, end=" ")

    for x in reversed(graph[node]):
        stack.append(x)
```

**3. Depth Limited Search — DLS**

```python
graph = {
    'A': ['B', 'C'],
    'B': ['D'],
    'C': ['E'],
    'D': [],
    'E': []
}

def dls(node, depth):

    print(node, end=" ")

    if depth == 0:
        return

    for x in graph[node]:
        dls(x, depth - 1)


dls('A', 1)
```

**4. Bidirectional Search**

```python
graph = {
    'A': ['B'],
    'B': ['C'],
    'C': ['D'],
    'D': ['E'],
    'E': []
}

start = 'A'
goal = 'E'

print("From Start:")
print(start, end=" ")

for x in graph:
    if x != start:
        print(x, end=" ")

print("\nFrom Goal:")
print(goal, end=" ")

for x in reversed(list(graph.keys())):
    if x != goal:
        print(x, end=" ")
```
