```python
facts = {
    "fever",
    "cough",
    "sore_throat"
}

rules = {
    "common_cold": ["cough", "sore_throat"],
    "flu": ["fever", "cough", "body_pain"],
    "throat_infection": ["fever", "sore_throat"],
    "respiratory_infection": ["fever", "cough"]
}


def forward_chaining(facts, rules):
    new_facts = True

    while new_facts:
        new_facts = False

        for conclusion, conditions in rules.items():
            if all(condition in facts for condition in conditions):
                if conclusion not in facts:
                    facts.add(conclusion)
                    print("Inferred:", conclusion)
                    new_facts = True

    return facts


def backward_chaining(goal, facts, rules):
    if goal in facts:
        return True

    if goal not in rules:
        return False

    for condition in rules[goal]:
        if not backward_chaining(condition, facts, rules):
            return False

    return True


print("Initial Facts:", facts)

print("\nForward Chaining:")
result = forward_chaining(facts.copy(), rules)
print("Final Facts:", result)

print("\nBackward Chaining:")
goal = "respiratory_infection"

if backward_chaining(goal, facts, rules):
    print(goal, "is supported by the facts")
else:
    print(goal, "cannot be concluded")
```
