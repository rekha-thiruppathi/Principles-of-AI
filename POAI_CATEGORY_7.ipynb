```python
variables = ['A', 'B', 'C', 'D']

domains = {
    'A': ['Red', 'Green', 'Blue'],
    'B': ['Red', 'Green', 'Blue'],
    'C': ['Red', 'Green', 'Blue'],
    'D': ['Red', 'Green', 'Blue']
}

constraints = [
    ('A', 'B'),
    ('A', 'C'),
    ('B', 'C'),
    ('B', 'D'),
    ('C', 'D')
]

assignment = {}


def is_valid(variable, value):
    for x, y in constraints:
        if x == variable and y in assignment:
            if assignment[y] == value:
                return False

        if y == variable and x in assignment:
            if assignment[x] == value:
                return False

    return True


def solve():
    if len(assignment) == len(variables):
        return True

    variable = next(v for v in variables if v not in assignment)

    for value in domains[variable]:
        if is_valid(variable, value):
            assignment[variable] = value

            if solve():
                return True

            del assignment[variable]

    return False


if solve():
    print("Map Coloring Solution:")
    for variable, value in assignment.items():
        print(variable, ":", value)
else:
    print("No solution exists")
```

```python
letters = ['S', 'E', 'N', 'D', 'M', 'O', 'R', 'Y']

domains = {letter: list(range(10)) for letter in letters}

assignment = {}

def is_valid(letter, digit):
    if digit in assignment.values():
        return False

    if letter in ['S', 'M'] and digit == 0:
        return False

    return True


def check_solution():
    if len(assignment) != 8:
        return False

    S = assignment['S']
    E = assignment['E']
    N = assignment['N']
    D = assignment['D']
    M = assignment['M']
    O = assignment['O']
    R = assignment['R']
    Y = assignment['Y']

    SEND = 1000*S + 100*E + 10*N + D
    MORE = 1000*M + 100*O + 10*R + E
    MONEY = 10000*M + 1000*O + 100*N + 10*E + Y

    return SEND + MORE == MONEY


def solve():
    if len(assignment) == len(letters):
        return check_solution()

    letter = letters[len(assignment)]

    for digit in domains[letter]:
        if is_valid(letter, digit):
            assignment[letter] = digit

            if solve():
                return True

            del assignment[letter]

    return False


if solve():
    print("Solution Found:")
    for letter in letters:
        print(letter, "=", assignment[letter])

    print("\nCryptarithmetic Equation:")

    SEND = int(''.join(str(assignment[x]) for x in 'SEND'))
    MORE = int(''.join(str(assignment[x]) for x in 'MORE'))
    MONEY = int(''.join(str(assignment[x]) for x in 'MONEY'))

    print(SEND, "+", MORE, "=", MONEY)
else:
    print("No solution found")
```

```python
import random

variables = ['A', 'B', 'C', 'D', 'E']

colors = ['Red', 'Green', 'Blue']

neighbors = {
    'A': ['B', 'C'],
    'B': ['A', 'C', 'D'],
    'C': ['A', 'B', 'D', 'E'],
    'D': ['B', 'C', 'E'],
    'E': ['C', 'D']
}

assignment = {}

for region in variables:
    assignment[region] = random.choice(colors)

print("Initial Assignment:")
print(assignment)


def count_conflicts(region, color):
    conflicts = 0

    for neighbor in neighbors[region]:
        if assignment[neighbor] == color:
            conflicts += 1

    return conflicts


def total_conflicts():
    conflicts = 0

    for region in variables:
        for neighbor in neighbors[region]:
            if assignment[region] == assignment[neighbor]:
                conflicts += 1

    return conflicts // 2


def min_conflicts(max_steps):

    for step in range(max_steps):

        if total_conflicts() == 0:
            return True

        conflicted = []

        for region in variables:
            if count_conflicts(region, assignment[region]) > 0:
                conflicted.append(region)

        region = random.choice(conflicted)

        minimum = float('inf')
        best_colors = []

        for color in colors:
            conflicts = count_conflicts(region, color)

            if conflicts < minimum:
                minimum = conflicts
                best_colors = [color]

            elif conflicts == minimum:
                best_colors.append(color)

        assignment[region] = random.choice(best_colors)

        print("Step", step + 1, ":", assignment)

    return total_conflicts() == 0


if min_conflicts(100):

    print("\nSolution Found!")
    print("Final Map Coloring:")

    for region in variables:
        print(region, ":", assignment[region])

    print("Total Conflicts:", total_conflicts())

else:
    print("\nNo solution found within the given steps.")
```
