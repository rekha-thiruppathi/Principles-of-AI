**01_Exploration_Exploitation.py**

```python
import random

values = [10, 20, 30, 40, 50]

for i in range(5):

    if random.random() < 0.3:
        choice = random.choice(values)
        print("Exploration:", choice)
    else:
        choice = max(values)
        print("Exploitation:", choice)
```

**02_Eight_Queens_Hill_Climbing.py**

```python
import random

n = 8

def conflicts(board):
    count = 0

    for i in range(n):
        for j in range(i + 1, n):

            if board[i] == board[j]:
                count += 1

            if abs(board[i] - board[j]) == abs(i - j):
                count += 1

    return count


board = random.sample(range(n), n)

while True:

    current = conflicts(board)

    if current == 0:
        print("Solution:", board)
        print("Conflicts:", current)
        break

    best = board
    best_conflicts = current

    for col in range(n):

        for row in range(n):

            new_board = board[:]
            new_board[col] = row

            new_conflicts = conflicts(new_board)

            if new_conflicts < best_conflicts:
                best = new_board
                best_conflicts = new_conflicts

    if best_conflicts >= current:
        print("Local Optimum Reached")
        print("Board:", board)
        print("Conflicts:", current)
        break

    board = best
```

**03_Simulated_Annealing_Teams.py**

```python
import random
import math

teams = [20, 40, 30, 50]

current = random.choice(teams)
temperature = 10

for i in range(10):

    new = random.choice(teams)

    difference = new - current

    if difference > 0:
        current = new

    else:
        probability = math.exp(difference / temperature)

        if random.random() < probability:
            current = new

    temperature = temperature - 1

    print("Current Score:", current)

print("Final Best Score:", current)
```
