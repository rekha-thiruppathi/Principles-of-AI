```python
def bayes_diagnosis(prior, symptoms, likelihoods):
    probability = prior

    for symptom in symptoms:
        probability *= likelihoods[symptom]

    return probability


def calculate_probability(prior, symptoms, likelihoods,
                          false_likelihoods):
    numerator = prior
    denominator = prior

    for symptom in symptoms:
        numerator *= likelihoods[symptom]
        denominator += 0

    for symptom in symptoms:
        denominator *= 1

    normal_probability = 1 - prior

    for symptom in symptoms:
        normal_probability *= false_likelihoods[symptom]

    denominator = numerator + normal_probability

    return numerator / denominator


diseases = {
    "Flu": {
        "prior": 0.30,
        "fever": 0.90,
        "cough": 0.80,
        "body_pain": 0.85,
        "false": {
            "fever": 0.20,
            "cough": 0.30,
            "body_pain": 0.25
        }
    },
    "Common Cold": {
        "prior": 0.40,
        "fever": 0.30,
        "cough": 0.85,
        "body_pain": 0.20,
        "false": {
            "fever": 0.70,
            "cough": 0.25,
            "body_pain": 0.80
        }
    }
}

symptoms = []

for symptom in ["fever", "cough", "body_pain"]:
    answer = input("Do you have " + symptom + "? (yes/no): ")
    if answer.lower() == "yes":
        symptoms.append(symptom)

print("\nMedical Diagnosis Results")

for disease, data in diseases.items():
    likelihoods = {
        symptom: data[symptom] for symptom in symptoms
    }

    probability = calculate_probability(
        data["prior"],
        symptoms,
        likelihoods,
        data["false"]
    )

    print(disease, ":", round(probability * 100, 2), "%")
```

```python
def battery_level_fuzzy(level):
    if level <= 20:
        return 1.0, 0.0, 0.0
    elif level <= 50:
        low = (50 - level) / 30
        medium = (level - 20) / 30
        return low, medium, 0.0
    elif level <= 80:
        medium = (80 - level) / 30
        high = (level - 50) / 30
        return 0.0, medium, high
    else:
        return 0.0, 0.0, 1.0


def temperature_fuzzy(temp):
    if temp <= 25:
        return 1.0, 0.0, 0.0
    elif temp <= 45:
        normal = (45 - temp) / 20
        warm = (temp - 25) / 20
        return normal, warm, 0.0
    else:
        return 0.0, 0.0, 1.0


battery = float(input("Enter Battery Level: "))
temperature = float(input("Enter Battery Temperature: "))

low, medium, high = battery_level_fuzzy(battery)
normal, warm, hot = temperature_fuzzy(temperature)

rules = {
    "Critical": max(low, hot),
    "Warning": max(medium, warm),
    "Good": min(high, normal)
}

print("\nFuzzy Logic Results")

for condition, value in rules.items():
    print(condition, ":", round(value, 2))

result = max(rules, key=rules.get)

print("\nBattery Condition:", result)

if result == "Critical":
    print("Action: Request safe charging or thermal inspection")
elif result == "Warning":
    print("Action: Monitor battery condition")
else:
    print("Action: Battery condition appears normal")
```

```python
def bayesian_update(prior, sensor_results, sensitivity,
                    false_positive):

    probability = prior
    not_probability = 1 - prior

    for result, accuracy, false_rate in zip(
        sensor_results, sensitivity, false_positive
    ):

        if result == "yes":
            probability *= accuracy
            not_probability *= false_rate

        else:
            probability *= (1 - accuracy)
            not_probability *= (1 - false_rate)

    return probability / (probability + not_probability)


prior = 0.20

sensors = ["Camera", "Radar", "Lidar"]

sensitivity = [0.90, 0.95, 0.98]
false_positive = [0.10, 0.05, 0.02]

results = []

for sensor in sensors:
    result = input(sensor + " detected obstacle? (yes/no): ")
    results.append(result.lower())

probability = bayesian_update(
    prior,
    results,
    sensitivity,
    false_positive
)

print("\nAutonomous Vehicle Analysis")
print("Obstacle Probability:",
      round(probability * 100, 2), "%")

if probability >= 0.80:
    print("Decision: High obstacle probability - request emergency stop")
elif probability >= 0.40:
    print("Decision: Uncertain - request cautious assessment")
else:
    print("Decision: Low obstacle probability - continue monitoring")
```
